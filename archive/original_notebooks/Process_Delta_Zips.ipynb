```python
from astropy.io import fits
import numpy as np
from scipy.io.wavfile import write as wav_write

fits_path = r"C:\Users\cmkua\Downloads\Delta\test_output_original_converter\first_fits\scam_0418_0704038328_265_ca0_scam02418_little_beach_________01p02.fits"

with fits.open(fits_path) as hdul:
    shot = hdul[6].data["Shot1"]   # <-- Second shot

audio = np.asarray(shot).astype(np.float32)

print("Samples:", len(audio))
print("Duration:", len(audio)/100000)

# Save as WAV
audio = audio - np.mean(audio)
audio /= np.max(np.abs(audio))

wav = np.int16(audio * 32767)

wav_write("test_shot2.wav", 100000, wav)

print("Saved test_shot2.wav")
```
